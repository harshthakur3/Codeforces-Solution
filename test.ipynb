```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import firwin, freqz

# Sampling frequency (Hz)
fs = 2000.0

# Cutoff frequencies (Hz)
low_cutoff = 200
high_cutoff = 400

# Filter order
numtaps = 101

# Design filters
lowpass = firwin(numtaps, cutoff=low_cutoff, fs=fs, pass_zero='lowpass')
highpass = firwin(numtaps, cutoff=low_cutoff, fs=fs, pass_zero='highpass')
bandpass = firwin(numtaps, cutoff=[low_cutoff, high_cutoff], fs=fs, pass_zero='bandpass')
bandstop = firwin(numtaps, cutoff=[low_cutoff, high_cutoff], fs=fs, pass_zero='bandstop')

# Frequency response plots
filters = {
    "Low-pass": lowpass,
    "High-pass": highpass,
    "Band-pass": bandpass,
    "Band-stop": bandstop
}

plt.figure(figsize=(10, 8))

for i, (name, coeffs) in enumerate(filters.items(), 1):
    w, h = freqz(coeffs, worN=8000, fs=fs)
    plt.subplot(2, 2, i)
    plt.plot(w, 20 * np.log10(np.abs(h)), 'b')
    plt.title(f"{name} Filter Frequency Response")
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude (dB)')
    plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sampling frequency
fs = 8000  # 8 kHz

# Duration of signal (seconds)
duration = 0.05  # 50 ms

# Time vector
t = np.linspace(0, duration, int(fs * duration), endpoint=False)

# Two example frequencies (Hz)
f1 = 697   # Low frequency tone
f2 = 1209  # High frequency tone

# Generate two sinusoidal tones
tone1 = np.sin(2 * np.pi * f1 * t)
tone2 = np.sin(2 * np.pi * f2 * t)

# Combine the tones
combined_tone = tone1 + tone2

# Plot the signals
plt.figure(figsize=(10, 6))

plt.subplot(3, 1, 1)
plt.plot(t, tone1)
plt.title('Tone 1 (f1 = 697 Hz)')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

plt.subplot(3, 1, 2)
plt.plot(t, tone2)
plt.title('Tone 2 (f2 = 1209 Hz)')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

plt.subplot(3, 1, 3)
plt.plot(t, combined_tone)
plt.title('Combined Dual Tone Signal (DTMF)')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

plt.tight_layout()
plt.show()
```
